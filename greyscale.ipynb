```python
import cv2
import os
import time
import re

# Define the classes
classes_to_capture = ['Eye Close', 'Open Mouth', 'Yawning', 'Normal']

# Function to extract numbers from filenames
def extract_number(filename):
    match = re.search(r'(\d+)', filename)
    return int(match.group()) if match else 0

# Prompt the user for the class to capture
class_input = int(input('Which Class do you want to capture?\n1. Eye Close\n2. Open Mouth\n3. Yawning\n4. Normal\nEnter number between 1-4:\n- '))

# Check class_input validity
if class_input < 1 or class_input > 4:
    print("Invalid input, exiting.")
    sys.exit()

# Print the class being captured
selected_class = classes_to_capture[class_input - 1]
print(f'Capturing {selected_class}')

# Ask for the duration of the video to be captured
capture_duration = int(input('How long should each video be (in seconds)?\nEg: if 10 sec enter 10\n- '))

frame_count = 0

# Main loop for capturing videos
while True:
    capture = input(f'Capture new frame for {selected_class}? (Y to capture, Q to quit)\n- ').lower()
    
    if capture == 'y':
        # Create a directory for the selected class if it doesn't exist
        directory = selected_class
        if not os.path.exists(directory):
            os.mkdir(directory)
            frame_count = 1
        else:
            # Get the max frame count from existing files
            frame_count = max([extract_number(filename) for filename in os.listdir(directory)] or [0]) + 1
        
        print(f'Capturing video {frame_count} for {selected_class}')
        
        # Open the default camera
        cam = cv2.VideoCapture(0)
        
        # Get the frame width and height
        frame_width = int(cam.get(cv2.CAP_PROP_FRAME_WIDTH))
        frame_height = int(cam.get(cv2.CAP_PROP_FRAME_HEIGHT))
        
        # Define the codec and create VideoWriter object
        fourcc = cv2.VideoWriter_fourcc(*'MJPG')
        out = cv2.VideoWriter(f'{directory}/{directory}_{frame_count}.avi', fourcc, 20.0, (frame_width, frame_height))
        
        start_time = time.time()
        
        # Capture video frames until the capture duration is reached
        while int(time.time() - start_time) < capture_duration:
            ret, frame = cam.read()
            if not ret:
                print("Failed to capture frame. Exiting.")
                break

            # Apply black-and-white filter (grayscale)
            gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

            # Optional: Night vision effect with green tint (uncomment if needed)
            # night_vision_frame = cv2.applyColorMap(gray_frame, cv2.COLORMAP_HOT)

            # Display grayscale or night vision frame
            cv2.imshow(f'Capturing {selected_class}', gray_frame)

            # Write the grayscale frame to the output file
            # For night vision effect, write night_vision_frame instead of gray_frame
            out.write(cv2.cvtColor(gray_frame, cv2.COLOR_GRAY2BGR))

            # Press 'q' to stop the capturing manually
            if cv2.waitKey(1) & 0xFF == ord('q'):
                break
        
        # Release the camera and writer objects
        cam.release()
        out.release()
        cv2.destroyAllWindows()

    elif capture == 'q':
        print("Exiting the capture loop.")
        break
    else:
        print("Invalid input. Please enter 'Y' to capture or 'Q' to quit.")
```
